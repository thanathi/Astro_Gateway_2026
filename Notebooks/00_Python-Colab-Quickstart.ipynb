## 4. วาดกราฟด้วย matplotlib

```python
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 300)                              # เวลา 0-10 วัน จำนวน 300 จุด
brightness = 1 + 0.3 * np.sin(2 * np.pi * t / 2.5)       # ความสว่างเปลี่ยนเป็นคาบ 2.5 วัน

plt.figure(figsize=(8, 3))
plt.plot(t, brightness, color="tab:orange")
plt.xlabel("time (days)")
plt.ylabel("brightness")
plt.title("A periodic signal")
plt.show()
```
